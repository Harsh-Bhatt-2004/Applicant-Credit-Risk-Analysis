## Working With The Model...

```python
# First Split data into training and testing sets.
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, train_size = 0.8, random_state = 42)
```

```python
# The KNN Mainly Used To Set The Point Which Is Derived From Which Cluster.... 
# Having Labelled Data Uses Supervised Machine Learning....

# Also To Find Distance Between Two Points Then :- √ (x2 - x1)² + (y2 - y2)²
```

```python
# Create a KNN classifier between Features and target data.
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model = KNeighborsClassifier(n_neighbors = 8)
```

```python
model.fit(x_train, y_train)
```

```python
# Display the test score.
```

```python
model.score(x_test, y_test)
```

```python
# Display the training score.
```

```python
model.score(x_train, y_train)
```

```python
# Print the accuracy score.
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, model.predict(x_test))
```

```python
# Try 1 to 14 k values for classifier.
```

```python
# It Means That We Need To Changes The Value Of "n_neighbours" And See What Are The 
# Training and Testing Scores...
```

```python
train_score_lis = []
test_score_lis = []

for i in range(1, 15):
    model = KNeighborsClassifier(n_neighbors = i)
    model.fit(x_train, y_train)

    train_score_lis.append(model.score(x_train, y_train))
    test_score_lis.append(model.score(x_test, y_test))
```

```python
# Display training and testing score for all the 1 to 14 k values.
```

```python
for i in range(0, 14):
    print("\n", i, train_score_lis[i])
```

```python
for i in range(0, 14):
    print("\n", i, test_score_lis[i])
```

```python
# We Can Also See The Data In The Chart Like Following...
```

```python
sns.lineplot(x = range(1,15), y = train_score_lis, marker = "*", label = "Train Score")
sns.lineplot(x = range(1,15), y = test_score_lis, marker = "o", label = "Test Score")
```
